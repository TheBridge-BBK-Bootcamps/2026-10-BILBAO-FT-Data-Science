## Ejercicio 1
Escribe una función que convierta números del 1 al 7 en nombres de los dias de la semana. La función constará de un único argumento numérico y una salida de tipo string

````python
def ejercicio_1(dia):
    dias_semana = {
        1: "Lunes",
        2: "Martes",
        3: "Miércoles",
        4: "Jueves",
        5: "Viernes",
        6: "Sábado",
        7: "Domingo"
    }
    return dias_semana.get(dia, "Día erróneo")
````

El uso de un diccionario asocia directamente cada clave (número) con su día correspondiente. El método .get(dia, "Día erróneo") busca el valor de la clave dia y, si la clave no existe (por ejemplo, con el número 8 o 100), devuelve por defecto la cadena "Día erróneo". Esto evita lanzar un error KeyError que interrumpiría la ejecución.

## Ejercicio 2
En el ejercicio 8 de flujos de control, creábamos una pirámide invertida de números desde el 5. Crea una función que replique el comportamiento de la pirámide, y utiliza un único parámetro de entrada de la función para determinar el número de filas de la pirámide.

````python
def ejercicio_2(tamaño: int = 5):
    lista = list(range(int(tamaño), 0, -1))
    while len(lista) > 0:
        print(*lista)
        lista.pop(0)
````

La función range(int(tamaño), 0, -1) genera la secuencia descendente de números enteros. La instrucción print(*lista) usa el operador de desempaquetado (*) para imprimir los elementos de la lista separados por espacios en lugar de mostrarlos dentro de corchetes. En cada iteración del bucle while, el método lista.pop(0) elimina el primer elemento (el número más alto restante), acortando la lista en cada paso hasta que queda vacía.

## Ejercicio 3
Escribe una función que compare dos números. La función tiene dos argumentos y hay tres salidas posibles: que sean iguales, que el primero sea  mayor que el segundo, o que el segundo sea mayor que el primero

````python
def ejercicio_3(num_1, num_2):
    if num_1 == num_2:
        out = "Son iguales"
    elif num_1 > num_2:
        out = "Primer número mayor que segundo"
    else:
        out = "Segundo número mayor que primero"
    return out
````

Se emplea la estructura de control if-elif-else para evaluar secuencialmente las tres condiciones excluyentes posibles. Si la primera comparación evalúa a True, retorna la cadena de igualdad; de lo contrario, comprueba cuál de las dos magnitudes es superior y devuelve el mensaje descriptivo correspondiente.   

## Ejercicio 4
Escribe una función que sea un contador de letras. En el primer argumento tienes que introducir un texto, y el segundo que sea una letra que contar. La función tiene que devolver un entero con el número de veces que aparece esa letra, tanto mayúscula, como minúscula

````python
def limpieza_texto(texto):
    dict_vocales = {'á': 'a', 'é': 'e', 'í': 'i', 'ó': 'o', 'ú': 'u'}
    texto = texto.lower()
    for clave, valor in dict_vocales.items():
        texto = texto.replace(clave, valor)
    return texto

def ejercicio_4(frase, letra):
    frase_limpia = limpieza_texto(frase)
    return frase_limpia.count(letra.lower())
````

Para hacer una búsqueda insensible a mayúsculas y acentos, la función auxiliar limpieza_texto convierte el texto completo a minúsculas con .lower() y sustituye las vocales acentuadas por sus equivalentes simples usando .replace(). Luego, el método nativo .count() busca la ocurrencia exacta de la letra dada (también convertida a minúscula).

## Ejercicio 6
Escribir una función que añada o elimine elementos en una lista. La función necesita los siguientes argumentos:
* lista: la lista donde se añadirán o eliminarán los elementos
* comando: "add" o "remove"
* elemento: Por defecto es None.

Tendrá como salida la lista

````python
def ejercicio_6(lista, comando, elemento=None):
    if comando == "add":
        lista.append(elemento)
    elif comando == "remove":
        try:
            lista.remove(elemento)
        except:
            print("El elemento no se encuentra en la lista")
    else:
        print("Comando no reconocido")
    return lista
````

````python
ejercicio_6([1,2,3],"add", 4)
````

## Ejercicio 8
Escribe un programa que obtenga el enésimo número de la [serie de Fibonacci](https://es.wikipedia.org/wiki/Sucesi%C3%B3n_de_Fibonacci). Tienes que crear una función recursiva con un único argumento.

```Python
fibonacci(2) = 1
fibonacci(4) = 3
fibonacci(5) = 5
fibonacci(8) = 21
```

fibonacci(n) = fibonacci(n-1) + fibonacci(n-2)

````python
def fibonacci(n):
    if n == 0 or n == 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)
````

## Ejercicio 9
Define en una única celda las siguientes funciones:
* Función que calcule el área de un cuadrado
* Función que calcule el área de un triángulo
* Función que calcule el área de un círculo

En otra celda, calcular el area de:
* Dos círculos de radio 10 + un triángulo de base 3 y altura 7
* Un cuadrado de lado = 10 + 3 círculos (uno de radio = 4 y los otros dos de radio = 6) + 5 triángulos de base = 2 + altura = 4

````python
import math
````

````python
def cuadrado(l, n=1):
    return round(n * l ** 2, 2)

def triangulo(b, a):
    return round(b * a / 2, 2)

def circulo(r):
    return round(math.pi * r ** 2, 2)
````

````python
print("Resultado 1:", 2 * circulo(10) + triangulo(3,7))
print("Resultado 2:", cuadrado(10) + circulo(4) + 2 * circulo(6) + 5 * triangulo(2,4))
````
